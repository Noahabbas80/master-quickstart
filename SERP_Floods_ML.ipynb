```python
import geopandas as gpd
import shapely.geometry
import json
from datetime import datetime

def parse_metadata(meta_path):
    with open(meta_path, 'r') as f:
        data = json.load(f)

    ems_code = data.get("ems_code", None)
    if not ems_code:
        raise KeyError("ems_code not found in metadata.")

    try:
        s2_date = datetime.strptime(data['s2_date'], '%Y-%m-%dT%H:%M:%S.%f%z')
    except ValueError:
        s2_date = datetime.strptime(data['s2_date'], '%Y-%m-%dT%H:%M:%S%z')

    bbox_dict = data.get("bounding box", None)
    if not bbox_dict:
        raise KeyError("bounding box not found in metadata.")

    bbox = [
        bbox_dict["west"],
        bbox_dict["south"],
        bbox_dict["east"],
        bbox_dict["north"]
    ]

    # extract the full AOI polygon as a GeoDataFrame
    if "area_of_interest_polygon" in data:
        polygon = shapely.geometry.shape(data["area_of_interest_polygon"])
        aoi = gpd.GeoDataFrame(geometry=[polygon], crs="EPSG:4326")
    else:
        aoi = None

    return ems_code, s2_date, bbox, aoi
```

```python
    from sentinelhub import SentinelHubRequest, DataCollection, MimeType, BBox, CRS, bbox_to_dimensions
import rasterio.transform
import numpy as np

def download_sentinel_image(bbox_coords, date, config, resolution=10):
    bbox = BBox(bbox=bbox_coords, crs=CRS.WGS84)
    size = bbox_to_dimensions(bbox, resolution=resolution)  # (width, height)

    request = SentinelHubRequest(
        evalscript=evalscript,
        input_data=[SentinelHubRequest.input_data(
            data_collection=DataCollection.SENTINEL2_L2A,
            time_interval=(date.strftime("%Y-%m-%d"), date.strftime("%Y-%m-%d")),
            mosaicking_order='leastCC'
        )],
        responses=[SentinelHubRequest.output_response('default', MimeType.TIFF)],
        bbox=bbox,
        size=size,
        config=config
    )

    image = request.get_data()[0]  # shape: (H, W, 3)

    # Compute affine transform manually
    west, south, east, north = bbox_coords
    width, height = size
    transform = rasterio.transform.from_bounds(west, south, east, north, width, height)
    crs = "EPSG:4326"

    return image, transform, crs
```

```python
def stretch(img, lower_percent=2, upper_percent=98):
    out = np.zeros_like(img, dtype=np.float32)
    for i in range(3):
        low = np.percentile(img[:, :, i], lower_percent)
        high = np.percentile(img[:, :, i], upper_percent)
        out[:, :, i] = np.clip((img[:, :, i] - low) / (high - low), 0, 1)
    return out

plt.imshow(stretch(image_test))
plt.title("Contrast-stretched RGB")
plt.axis('off')
plt.show()
```

```python
from rasterio.features import rasterize
import geopandas as gpd
import numpy as np

def rasterize_flood_mask(geojson_path, out_shape, transform):
    gdf = gpd.read_file(geojson_path)

    # Only keep features labeled as "flood"
    flood_gdf = gdf[gdf["class"] == "flood"]

    if flood_gdf.empty:
        print(f"No flood polygons found in: {geojson_path}")
        return np.zeros(out_shape, dtype='uint8')

    # Rasterize the flood polygons
    mask = rasterize(
        [(geom, 1) for geom in flood_gdf.geometry],
        out_shape=out_shape,
        transform=transform,
        fill=0,
        dtype='uint8'
    )
    return mask
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class UNet(nn.Module):
    def __init__(self, in_channels=3, out_channels=1):
        super().__init__()
        def CBR(in_ch, out_ch):
            return nn.Sequential(
                nn.Conv2d(in_ch, out_ch, 3, padding=1),
                nn.BatchNorm2d(out_ch),
                nn.ReLU(inplace=True),
            )

        self.enc1 = CBR(in_channels, 64)
        self.enc2 = CBR(64, 128)
        self.enc3 = CBR(128, 256)

        self.pool = nn.MaxPool2d(2, 2)

        self.dec2 = CBR(256 + 128, 128)
        self.dec1 = CBR(128 + 64, 64)

        self.up2 = nn.ConvTranspose2d(256, 128, 2, stride=2)
        self.up1 = nn.ConvTranspose2d(128, 64, 2, stride=2)

        self.final = nn.Conv2d(64, out_channels, kernel_size=1)

    def forward(self, x):
        e1 = self.enc1(x)
        e2 = self.enc2(self.pool(e1))
        e3 = self.enc3(self.pool(e2))

        d2 = self.up2(e3)
        d2 = self.dec2(torch.cat([d2, e2], dim=1))
        d1 = self.up1(d2)
        d1 = self.dec1(torch.cat([d1, e1], dim=1))

        return torch.sigmoid(self.final(d1))
```

```python
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader

class FloodDataset(Dataset):
    def __init__(self, img_path, mask_path):
        self.image = np.load(img_path).astype(np.float32).transpose(2, 0, 1) / 3000.0
        self.mask = np.load(mask_path).astype(np.float32)[None, :, :]  # add channel
    def __len__(self):
        return 1
    def __getitem__(self, idx):
        return torch.tensor(self.image), torch.tensor(self.mask)

dataset = FloodDataset("image_rgb.npy", "flood_mask.npy")
loader = DataLoader(dataset, batch_size=1)

model = UNet()
optimizer = optim.Adam(model.parameters(), lr=1e-3)
loss_fn = nn.BCELoss()

for epoch in range(10):
    for img, mask in loader:
        pred = model(img)
        loss = loss_fn(pred, mask)
        loss.backward()
        optimizer.step()
        optimizer.zero_grad()
    print(f"Epoch {epoch}, Loss: {loss.item():.4f}")
```
